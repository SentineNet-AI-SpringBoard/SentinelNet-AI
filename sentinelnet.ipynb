```python
from scipy import stats
import numpy as np
z = np.abs(stats.zscore(X))
outliers = X[z > 3]
outliers.shape
```

##Interquartile Range (IQR) method to detect outliers in the features of dataset X

```python
Q1 = X.quantile(0.25)
Q3 = X.quantile(0.75)
IQR = Q3 - Q1

# Mark rows as outliers if any feature lies outside [Q1-1.5*IQR, Q3+1.5*IQR]
outliers = ((X < (Q1 - 1.5 * IQR)) | (X > (Q3 + 1.5 * IQR))).any(axis=1)

print(outliers)
```

```python
print("Number of outliers detected by the IQR method:", outliers.sum())
```
